```python
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("IPH_robos_2015_2025_tecmty.xlsx")
df.head()
```

```python
# Quiero que en la columna HORA las horas que sean 99 se cambien a NaN
df['HORA'] = df['HORA'].replace(99, pd.NA)
```

```python
# Saca la media de la columna HORA
mean_hour = df['HORA'].mean()
print("Media de la columna HORA:", mean_hour)
```

```python
# Dame una fila aleatoria del DataFrame
random_row = df.sample(n=1)
print("Fila aleatoria del DataFrame:")
print(random_row)
```

```python
# Dame la desviación estándar de la columna HORA
std_dev_hour = df['HORA'].std()
print("Desviación estándar de la columna HORA:", std_dev_hour)
```
